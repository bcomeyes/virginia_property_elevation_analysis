```python
import pandas as pd
from pyproj import Transformer

r = pd.read_csv("output/regions_20ft.csv")
tf = Transformer.from_crs(32618, 4326, always_xy=True)
r["lon"], r["lat"] = tf.transform(r["x"].values, r["y"].values)

c = r[(r.county == "Carteret") & r.pad_fits].sort_values("pad_acres", ascending=False)
print(c[["acres", "mean_ft", "max_ft", "pad_acres", "dist_water_km", "lat", "lon"]].to_string(index=False))
```
